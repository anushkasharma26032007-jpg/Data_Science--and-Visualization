```python
my_tuple = (10, 20, 30, 40, 50)

print(my_tuple[:3])    # (10, 20, 30)
print(my_tuple[2:])    # (30, 40, 50)
print(my_tuple[1:4])   # (20, 30, 40)
print(my_tuple[:])     # (10, 20, 30, 40, 50)
print(my_tuple[::-1])  # (50, 40, 30, 20, 10)
```

```python
#Tuple is immutable
my_tuple = (10, 20, 30, 40, 50)
my_tuple[0] = 100
```

```python
## add an item in a set
student = {"Amit","Radha","Kiran","kirti"}
student.add("Priya")
print(student)
student.update(["Rani","Ramesh"])
print(student)
student.remove("Ramesh")
print(student)
```

```python
students = {"Amit","Radha","Kiran","kirti"}
students.add("Priya")
print(students)
students.update(["Rani","Ramesh"])
print(students)
students.discard("Ramesh")
print(students)
```

```python
students={"priya","GITA","Karan"}
students.remove("Ramesh")
print(students)
```

```python
students = {"Amit","Radha","Kiran","kirti"}
students.add("Priya")
print(students)
# Discard does not give error if an elemnet is not present in your set if remove or discard
## it is not give an errir for that element
students.discard("Ramesh")
print(students)
```

```python
# create a set and clear list of item
student = {"Amit","Radha","Kiran","kirti"}
student.clear()
print(student)
```

```python
## set operation
python_students = {"Amit","neha","sanjay","manish"}
java_student = {"dhilip","anish","neha","karan"}

# for union operation
result = python_students.union(java_student)
print(result)
## for intersection operation using set
result = python_students.intersection(java_student)
print(result)
## for intersection operation using &
result = python_students & java_student
print(result)

# Differnce with two  set
  result = python_students.differnce(java_students)
  print(result)
```
